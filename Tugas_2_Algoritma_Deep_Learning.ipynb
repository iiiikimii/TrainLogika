# Membuat Data Training

```python
# Data Input
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)

# Target masing - masing logika gerbang logika
Y = {
    "AND": np.array([[0], [0], [0], [1]], dtype=float),
    "OR": np.array([[0], [1], [1], [1]], dtype=float),
    "XOR": np.array([[0], [1], [1], [0]], dtype=float),
    "NAND": np.array([[1], [1], [1], [0]], dtype=float),
    "NOR": np.array([[1], [0], [0], [0]], dtype=float),
    "XNOR": np.array([[1], [0], [0], [1]], dtype=float)
}

print("Input:")
print(X)

for nama, target in Y.items():
  print(f"\nTarget {nama}:")
  print(target.ravel())
```

# Membuat Fungsi Aktivasi dan Training

```python
def sigmoid(x):
    x = np.clip(x, -500, 500)
    return 1 / (1 + np.exp(-x))
```

# Membuat Fungsi Training Neural Network

```python
def train(X, y, epochs=20000, learning_rate=0.1, seed=42):
    rng = np.random.default_rng(seed)

    # Inisialisasi bobot dan bias
    W1 = rng.normal(0, 0.5, (2, 8))
    b1 = np.zeros((1, 8))

    W2 = rng.normal(0, 0.5, (8, 1))
    b2 = np.zeros((1, 1))

    losses = []
    jumlah_data = len(X)

    for epoch in range(epochs):

        # Forward propagation
        z1 = X @ W1 + b1
        a1 = np.tanh(z1)

        z2 = a1 @ W2 + b2
        a2 = sigmoid(z2)

        # Menghitung binary cross-entropy loss
        eps = 1e-8
        loss = -np.mean(
            y * np.log(a2 + eps)
            + (1 - y) * np.log(1 - a2 + eps)
        )

        # Backpropagation
        dz2 = (a2 - y) / jumlah_data
        dW2 = a1.T @ dz2
        db2 = np.sum(dz2, axis=0, keepdims=True)

        dz1 = (dz2 @ W2.T) * (1 - a1 ** 2)
        dW1 = X.T @ dz1
        db1 = np.sum(dz1, axis=0, keepdims=True)

        # Update bobot dan bias
        W2 -= learning_rate * dW2
        b2 -= learning_rate * db2
        W1 -= learning_rate * dW1
        b1 -= learning_rate * db1

        # Simpan loss setiap 100 epoch
        if (epoch + 1) % 100 == 0:
            losses.append(loss)

        # Tampilkan progres setiap 1000 epoch
        if (epoch + 1) % 1000 == 0:
            print(
                f"Epoch {epoch + 1:5d}/{epochs} "
                f"| Loss: {loss:.6f}"
            )

    model = (W1, b1, W2, b2)

    return model, losses
```

# Membuat fungsi prediksi

```python
def predict(X, model):
    W1, b1, W2, b2 = model

    z1 = np.dot(X, W1) + b1
    a1 = np.tanh(z1)

    z2 = np.dot(a1, W2) + b2
    output = sigmoid(z2)

    return output
```

# Training Seluruh Gerbang Logika

```python
models = {}
history = {}

for i, (nama, target) in enumerate(Y.items()):
    print(f"\n{'=' * 40}")
    print(f"Training Gerbang {nama}")
    print(f"{'=' * 40}")

    model, losses = train(
        X,
        target,
        epochs=20000,
        learning_rate=0.1,
        seed=42 + i
    )

    models[nama] = model
    history[nama] = losses

print("\nSemua proses training selesai!")
```
